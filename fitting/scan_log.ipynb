```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
cost,tzmultvalue,invtau_iso,invtau_aniso = np.loadtxt("fit_logs_22perc_35K.txt", skiprows=1,unpack=True,delimiter=",")
```

```python
%matplotlib inline
```

```python
indices_to_keep = cost < 0.06
color = cost[indices_to_keep]
x = invtau_aniso[indices_to_keep]
y = invtau_iso[indices_to_keep]
dx,dy = np.gradient(x),np.gradient(y)


plt.scatter(x,y,c=color)
plt.quiver(x, y, dx, dy, angles='xy', scale_units='xy', scale=1)
plt.show()
```
